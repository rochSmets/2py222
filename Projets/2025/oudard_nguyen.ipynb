## I. Résolution analytique

Pour une modélisation analytique du booster en chute libre avec un **angle d'attaque $\theta$ nul** (le booster est parfaitement aligné avec sa vitesse de descente, donc pas de portance), nous allons appliquer le **Principe Fondamental de la Dynamique (PFD)** dans le référentiel terrestre supposé galiléen.  

Nous faisons les hypothèses suivantes :

1. $\theta$ = 0. Le booster est en position verticale, donc il ne subit aucune force de portance ($ F_L = 0 $) ;
2. Nous ne considérons que le booster en état de chute libre : la poussée des réacteurs Raptor est ignorée ($F_T = 0 $) ;
3. La trajectoire est uniquement selon l'axe $z$ (orienté vers le haut) ;
4. La pseudo-force de Coriolis et la poussée d'Archimède sont négligées ;
5. Il n'y a pas de vent : la vitesse relative est égale à la vitesse de descente ;
6. $\rho = cte$, de même que la surface effective $A_{eff}$

Ainsi, nous pouvons effectuer un bilan des forces en présence :

- **Poids** : $ \vec{F}_g = -mg \vec{z} $ (vers le bas) ;
- **Traînée aérodynamique** : $ \vec{F}_D = \frac{1}{2} \rho v^2 C_D $A_{eff}$ \vec{z} $ (toujours opposée au mouvement, donc vers le haut si le booster descend).  

En appliquant le PFD, nous sommes en mesure d'obtenir l'équation du mouvement :

$$ m \frac{dv}{dt} = \sum F_z \implies m \frac{dv}{dt} = F_D + F_g \implies m \frac{dv}{dt} = \frac{1}{2} \rho v^2 C_D A - mg $$

L'équation résultante est l'équation différentielle non-linéaire du premier ordre 

$$ \frac{dv}{dt} = \frac{\rho C_D A}{2m} v^2 - g $$

de la forme :

$$ \frac{dv}{dt} = k v^2 - g $$  

où $ k = \frac{\rho C_D A}{2m} $ est une constante qui dépend de la densité de l'air, du coefficient de traînée, de la surface de référence et de la masse. Cette équation différentielle est à variables séparables :

$$ \frac{dv}{k v^2 + g} = -dt $$

Nous intégrons des deux côtés :

$$ \int \frac{dv}{k v^2 + g} = - \int dt $$  

Pour intégrer le membre de gauche, nous pouvons le factoriser de manière appropriée en utilisant les fractions partielles. En posant $ A = \sqrt{\frac{g}{k}} $, alors le dénominateur devient :

$$kv^2 - g = k(v^2 - A^2)= k(v - A)(v + A)$$

Nous écrivons alors :

$$\frac{1}{k(v - A)(v + A)} = \frac{B}{v - A} + \frac{C}{v + A}$$

En résolvant cette décomposition en fractions partielles, nous obtenons :

$$\frac{1}{k(v^2 - A^2)} = \frac{1}{2Ak} \left( \frac{1}{v - A} - \frac{1}{v + A} \right)$$

Enfin, nous intégrons les deux côtés :

$$\int \frac{1}{kv^2 - g} \, dv = \int \frac{1}{2Ak} \left( \frac{1}{v - A} - \frac{1}{v + A} \right) dv = \frac{1}{2Ak} \ln\left| \frac{v - A}{v + A} \right| + C$$

Donc :

$$\frac{1}{2Ak} \ln\left| \frac{v - A}{v + A} \right| = t + C$$

Nous exprimons maintenant $v(t)$ :

$$\ln\left| \frac{v - A}{v + A} \right| = 2Ak(t + C)$$

Nous posons $C_1 = e^{2AkC}$, d'où :

$$\left| \frac{v - A}{v + A} \right| = C_1 e^{2Akt}$$

Or, ici, nous considérons des vitesses positives : nous pouvons donc retirer la valeur absolue de l'expression.

$$\frac{v - A}{v + A} = C_1 e^{2Akt}$$

D'où, enfin :

$$v - A = C_1 e^{2Akt}(v + A)$$

$$v - C_1 e^{2Akt} v = C_1 e^{2Akt} A + A$$

$$v(1 - C_1 e^{2Akt}) = A(1 + C_1 e^{2Akt})$$

$$v = A \cdot \frac{1 + C_1 e^{2Akt}}{1 - C_1 e^{2Akt}}$$

Nous définissons la constante $ C_1 $ via la condition initiale $ v(0) = 0 $ (vitesse initiale nulle à l'apogée, seul cas où $v(t)$ prend une expression simple). Alors :

$$0 = A \cdot \frac{1 + C_1}{1 - C_1} \implies C_1 = -1$$

Nous remplaçons $ C_1 $ dans la Solution

$$v(t) = A \cdot \frac{1 - e^{2Akt}}{1 + e^{2Akt}} \implies v(t) = A \cdot \frac{e^{-Akt} - e^{Akt}}{e^{-Akt} + e^{Akt}}$$

et

$$v(t) = A \cdot \frac{-2 \sinh(Akt)}{2 \cosh(Akt)} = -A \tanh(Akt)$$

En remplaçant $ A = \sqrt{\frac{g}{k}} $ et $ k = \frac{\rho C_D A}{2m} $ :

$$v(t) = -\sqrt{\frac{g}{k}} \tanh\left( \sqrt{gk} \, t \right)$$

$$\boxed{v(t) = -\sqrt{\frac{2mg}{\rho C_D A}} \tanh\left( \sqrt{\frac{\rho C_D A g}{2m}} \, t \right)}$$


Cependant, cette solution est à prendre avec des pincettes, car beaucoup d'hypothèses sont faites. Finalement, en admettant les hypothèses, nous obtenons une expression que l'on affiche par rapport à la trajectoire de notre booster dans le cadre de notre simulation numérique, **en chute libre** :

## II. Résolution numérique

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from enum import Enum
````

Dans la section précédente, nous avons constaté toutes les contraintes imposées par la résolution analytique : sans ces différentes hypothèses, il n'existe pas de solution exacte à l'équation différentielle, donc nous devons procéder à une résolution numérique. En particulier, nous souhaitons nous passer des hypothèses de $\rho = cte$, absolument pas réaliste, de $C_D = cte$, de $v_0 = 0$, car dans les relevés de télémétrie officiels, le booster ne débute pas sa descente à vitesse nulle. Nous souhaitons enfin intégrer une dimension angulaire à la trajectoire, cette dernière n'étant absolument pas verticale. Ainsi, la surface effective $A_{eff}$ serait modifiée et non plus constante, autre hypothèse de la résolution analytique.

Nous avons choisi de modéliser le problème en **Programmation Orientée Objet** sur Python, bien plus flexible et puissante qu'en définissant une succession de fonctions. Pour cette modélisation, nous avons découpé chaque partie matérielle (Raptors, grid fins,...) et chaque partie indépendante liée au paramètres physiques (densité, pression,...) sous la forme de classes séparées.

### 1) Modélisation de l'atmosphère

L'atmosphère terrestre, implémentée dans la classe `Atmosphere`, est le coeur thermodynamique de la dynamique de vol du booster : à chaque pas d'intégration $dt$, elle doit être en mesure de fournir des valeurs continues de température et de pression, utiles au calcul de la masse volumique de l'air $\rho$, de la vitesse du son $c$, ou des forces de traînée et de portance dans la classe `Aerodynamics`.
L'évolution de la température, influant sur la pression atmosphérique et donc sur la densité, subit des discontinuités de pente entre les différentes couches d'atmosphère : la trajectoire s'étendant sur plusieurs dizaines de kilomètres d'altitude, un simple profil exponentiel de densité ne suffit ainsi plus pour dimensionner correctement les forces aérodynamiques à l'oeuvre durant la descente. Afin de palier ces spécificités, notre code adopte donc le modèle **International Standard Atmosphère (ISA)**, validé expérimentalement jusqu'à 85km d'altitude, et qui divise l'atmosphère en sept couches isothermes ou à gradients thermiques linéaires.

De sorte à ce que l'erreur relative calculée sur la densité soit inférieure à 2% pour toute altitude, ce modèle nous impose les hypothèses suivantes :

1. L'atmosphère est sèche, c'est-à-dire un mélange de gaz parfaits dénué d'humidité, et de composition constante : elle est constituée de deux molécules diatomiques, le dioxygène et le diazote, aux proportions respectives de 21% et 79%. Sa masse molaire moyenne est donc approximativement $\bar{M} = 28.84$ g/mol. Ce résultat nous permet de déterminer la constante spécifique des gaz pour l'air, qui nous sera utile lorsque nous souhaiterons calculer la densité de l'air et la vitesse du son à une altitude donnée, et de la fixer à $R = \frac{R*}{\bar{M}} = 282.1$ J/kg/K, où $R*=8.134$ J/mol/K est la constante des gaz parfaits ;
 
2. Comme l'atmosphère étudiée ne se situe qu'à quelques dizaines de kilomètres de la surface terrestre ($\frac {65km}{6400km}$, $\frac {altitude}{rayon \ de \ la \ Terre}$ : variation de l'ordre de **1%**), la norme de l'accélération de pesanteur est supposée constante, induisant une gravité uniforme. Nous fixerons alors cette constante égale à $g = 9.81$ m/s² ;  

3. L'atmosphère étant constituée d'un gaz parfait, nous sommes en mesure, à partir de l'équation des gaz parfaits $PV = n{R^*}\ T$, de démontrer la relation : 

$$\rho = \frac{P}{R^*T}$$

Attention, ici R* n'est pas la constante des gaz parfaits, et est spécifique à chaque gaz.

#### Température

La classe `Atmosphere` pose les bases de ce modèle atmosphérique à partir de la fonction `get_temperature(h)`, qui implémente les formulations suivantes (toutes les températures T sont données en Kelvin K):

- 0 – 11 000km (Troposphère) : $T = T_0 - 0.0065h$
- 11 000 – 20 000km (Tropopause) : $T = 216.65$
- 20 000 – 32 000km (Stratosphère) : $T = 196.65 + 0.001h$
- 32 000 – 47 000km (Stratosphère) : $T = 139.05 + 0.0028h$
- 47 000 – 51 000km (Stratopause) : $T = 270.65$ 
- 51 000 – 71 000km (Mésosphère) : $T = 413.45 - 0.0028h$
- 71 000 – 85 000km (Mésosphère) : $T = 356.65 - 0.0020h$

````python
class Atmosphere:
    def __init__(self):
        self.p0 = 101325.0   # Pa
        self.R  = 287.053    # J/(kg·K)
    
    def get_temperature(self, h):
        # Kelvin
        if h<0:
            h=0 
        if h<=11000:
            return 288.15 - 6.5*h/1000
        elif h<=20000:
            return 216.65
        elif h<=32000:
            return 196.65 + h/1000
        elif h<=47000:
            return 139.05 + 2.8*h/1000
        elif h<=51000:
            return 270.65
        elif h<=71000:
            return 413.45 - 2.8*h/1000
        else:
            return 356.65 - 2.0*h/1000
````

#### Pression

La fonction `get_pressure()` combine, pour chaque couche, l'équilibre hydrostatique $\frac{\mathrm{d}P}{\mathrm{d}h} = -\rho g_0$ à l'équation d'état des gaz parfaits :

- Au sein des couches à gradient $\beta$ non nul, nous obtenons $P(h) = P_b\;\Bigl(\tfrac{T_b}{T_b+\beta\,(h-h_b)}\Bigr)^{\tfrac{g}{R\beta}}$, où $b$ désigne la base de la couche ;

- Au sein des couches isothermes, c'est-à-dire telles que $\beta = 0$, alors $P(h) = P_b \exp[{-\frac{g}{RT_b}(h-h_b)}]$.

Voici le détail des calculs :

**Cas à gradient non nul**
 
$$ \frac{dP}{dh} = -\left(\frac{P}{RT}\right) g_0 $$
  
$$ \frac{dP}{P} = -\frac{g_0}{R}\frac{dh}{T(h)} $$

 
Avec $T(h) = T_b + \beta(h - h_b)$, donc $dT = \beta dh$ ou $dh = dT/\beta$

$$\int_{P_b}^{P} \frac{dP'}{P'} = -\frac{g_0}{R} \int_{T_b}^{T} \frac{dT'}{\beta T'}$$

$$\ln\left(\frac{P}{P_b}\right) = -\frac{g_0}{R\beta} \ln\left(\frac{T}{T_b}\right)$$

$$\frac{P}{P_b} = \left(\frac{T}{T_b}\right)^{-g_0/(R\beta)}$$

d'où 

$$P(h) = P_b \left(\frac{T_b + \beta(h - h_b)}{T_b}\right)^{-g_0/(R\beta)}$$


**Cas isotherme ($\beta = 0$)**

 
$$\frac{dP}{dh} = -\frac{P}{RT_b}g_0$$

  
$$\frac{dP}{P} = -\frac{g_0}{RT_b}dh$$


$$\int_{P_b}^{P} \frac{dP'}{P'} = -\frac{g_0}{RT_b} \int_{h_b}^{h} dh'$$

 
$$\ln\left(\frac {P}{P_b}\right) = -\frac{g_0}{RT_b}(h - h_b)$$

  
$$P(h) = P_b \exp\left[-\frac{g_0}{RT_b}(h - h_b)\right]$$

d'où

$$P(h) = P_b e^{-\frac{g_0}{RT_b}(h - h_b)}$$

<center><img src="isa.svg" alt="Modèle ISA" style="width:200px; height:600px;"></center>
<center>Figure 7 : Evolution de la température et de la pression en fonction de l'altitude. <br> Ces courbes suivent le modèle ISA.</center>

````python
def get_pressure(self, h):
    if h<0:
        h=0
    if h<=11000:
        return self.p0*(288.15/(288.15-6.5*h/1000))**(34.1632/-6.5)
    elif h<=20000:
        p11 = self.p0*(288.15/216.65)**(34.1632/-6.5)
        return p11 * np.exp(-34.1632*(h/1000-11)/216.65)
    elif h<=32000:
        p20= self.get_pressure(20000)
        T20=216.65
        return p20*(T20/self.get_temperature(h))**34.1632
    elif h<=47000:
        p32=self.get_pressure(32000)
        T32=self.get_temperature(32000)
        return p32*(T32/self.get_temperature(h))**(34.1632/2.8)
    elif h<=51000:
        p47=self.get_pressure(47000)
        return p47*np.exp(-34.1632*(h/1000-47)/270.65)
    elif h<=71000:
        p51=self.get_pressure(51000)
        T51=self.get_temperature(51000)
        return p51*(T51/self.get_temperature(h))**(34.1632/-2.8)
    else:
        p71=self.get_pressure(71000)
        T71=self.get_temperature(71000)
        return p71*(T71/self.get_temperature(h))**(34.1632/-2)

Atmosphere.get_pressure = get_pressure
````

#### Densité et vitesse du son

Une fois $T(h)$ et $P(h)$ obtenus, la densité est directement calculée par la fonction `density`, et ce à partir de la formule :  

$$\rho(h) = \frac{P(h)}{R\,T(h)}$$

La vitesse du son, qui intervient plus tard pour le calcul du nombre de Mach, est déterminée dans la fonction `speed_sound` par le biais de la relation :  

$$c(h) = \sqrt{\gamma\,R\,T(h)} \qquad \gamma = 1.4$$

````python
def density(self, h):
    T=self.get_temperature(h)
    P=self.get_pressure(h)
    return P/(self.R*T)
    
def sound_speed(self, h):
    gamma=1.4
    T=self.get_temperature(h)
    return np.sqrt(gamma*self.R*T)

Atmosphere.density = density
Atmosphere.sound_speed = sound_speed
````

### 2) Quaternions

Lors de sa phase de descente aérodynamique, le booster contrôle sa trajectoire par le biais de son orientation : la dynamique du point matériel n'étant pas un outil adapté à ce contexte, nous devons obligatoirement tenir compte des 3 degrés de liberté liés aux rotations du booster dans la description de son état. Nous aurions pu utiliser les angles d'Euler ($\psi,\theta,\phi$), plus connus en aéronautique sous les noms de lacet, roulis et tangage, ou des matrices de rotation, mais ces deux méthodes présentent des limites non négligeables : la première est certes intuitive, mais peut entraîner la perte d'un degré de liberté pour certaines orientations (verrouillage de Cardan), tandis que la seconde implique des temps de calcul bien plus importants. Nous avons donc décidé d'implémenter une autre méthode, basée sur un groupe mathématique : celle des quaternions, compacte et exempte de singularités.

#### Structure algébrique

Un quaternion est un nombre hypercomplexe s’écrivant comme la somme d’une partie réelle et d'une partie vectorielle :

$$q = a + b\mathbf{i} + c\mathbf{j} + d\mathbf{k} = (a, b, c, d)$$

où $(\mathbf{i}, \mathbf{j}, \mathbf{k})$ sont des unités imaginaires obéissant aux règles de Hamilton :

$$\mathbf{i}^2 = \mathbf{j}^2 = \mathbf{k}^2 = \mathbf{ijk} = -1$$

Ces règles définissent une structure algébrique non commutative, mais associative, permettant de composer les quaternions selon un produit bien défini : le produit de Hamilton. En notation vectorielle, si nous notons $q = (a, b)$ et $r = (e, f)$, avec $b, f \in \mathbb{R}^3$, ce produit s’écrit :

$$q\!*r = (a_e - b \cdot f \; , \; af + eb + b \times f)$$

La méthode `mul(q, r)` implémente précisément cette opération, qui joue un rôle central dans la composition successive des rotations, grâce à son associativité.

#### Quaternion et rotation

Nous pouvons associer un quaternion **unitaire**, c'est-à-dire pour lequel $\lVert q\rVert = 1$, à une rotation $R$ d'un angle $\theta$ autour d'un axe unitaire $\vec{h}$ :

$$q = \cos(\frac{\theta}{2}) + (\vec{h} \sin(\frac{\theta}{2}))$$

Le vecteur $\vec{v'}$, résultant de la rotation de $\vec{v}$ par $R$, s'écrit alors sans matrice :

$$\boxed{\vec{v'} = q \vec{v} q^\star}$$

avec $q^\star=(a,-b,-c,-d)$ le quaternion conjugué. La fonction `rot(v,q)` permet donc de calculer les nouvelles coordonnées de chaque vecteur après une rotation quelconque.

#### Normalisation

Comme établi ci-dessus, nous associons toute relation à un quaternion **unitaire**, donc de norme 1. Or, dans une intégration numérique, les arrondis effectués par Python induisent toujours une dérive dans cette normalité : sans correction adequate, la rotation effective s'éloigne peu à peu de la réalité, contaminant ainsi la modélisation en injectant des vitesses angulaires fantômes. La fonction `norm(q)`, peu coûteuse en temps de calcul mais cruciale pour éviter les nuisances, applique donc :

$$q \leftarrow \frac{q}{\lVert q\rVert}$$

````python
class Quat:
    """
    Fonctions utilitaires pour quaternions :
    - mul : multiplication q * r
    - rot : rotation d'un vecteur v par q
    - norm: normalisation d'un quaternion
    """
    @staticmethod # Déclare la méthode comme statique, évitant l'utilisation de self 
                  # et permettant un appel direct depuis la classe.`
    def mul(q,r):
        a,b,c,d = q
        e,f,g,h = r
        return np.array([
            a*e-b*f-c*g-d*h,
            a*f+b*e+c*h-d*g,
            a*g-b*h+c*e+d*f,
            a*h+b*g-c*f+d*e
        ])
    @staticmethod
    def rot(v,q):
        """
        Rotate vector v by quaternion q:
        q * [0,v] * q_conj
        """
        qv = np.hstack(([0.],v))
        qc = np.array([q[0],-q[1],-q[2],-q[3]])
        return Quat.mul(Quat.mul(q,qv),qc)[1:]
    @staticmethod
    def norm(q):
        """Renvoie le quaternion q normalisé."""
        return q/np.linalg.norm(q)
````

Nous sommes désormais en mesure, maintenant que nous avons introduit la méthode d'orientation du booster dans l'espace, de présenter comme nous propageons son état de vol, c'est-à-dire le vecteur regroupant sa position, sa vitesse, ses quaternions, sa vitesse angulaire et sa masse. 

### 3) Intégrateur de Runge–Kutta–Fehlberg (RKF-45) 

Pour ce faire, il est nécessaire d'employer un schéma d'intégration offrant à la fois une précision élevée et un temps de calcul modéré. En effet, nous devons résoudre numériquement des equations différentielles ne présentant pas de solutions analytiques (cf paragraphe sur la résolution analytique). Notre choix s'est donc porté sur la méthode de **Runge-Kutta-Fehlberg d'ordre 4**, dite RKF-45, correspondant à une version adaptative de l'intégrateur RK4 simple : en effet, quand la méthode RK4 progresse à pas de temps fixe, son homologue RKF-45 calcule un terme d'ordre supérieur afin d'estimer une erreur locale, permettant alors d'adapter la durée du pas.

L’idée générale est d’évaluer la dérivée $f(t,X)$ six fois selon des combinaisons de l’état $X$ et du pas de temps $\Delta t$, chaque évaluation produisant un "incrément de pente" $k_i$. La fonction `rkf45_step` enchaîne ainsi, à partir du tableau publié par Felhberg en 1969 : 

$$
\begin{aligned}
k_1 &= f(t, X),\\
k_2 &= f(t + \frac{1}{4} \Delta t, X + \frac{1}{4} \Delta t k_1),\\
k_3 &= f(t + \frac{3}{8} \Delta t, X + \Delta t (\frac{3}{32} k_1 + \frac{9}{32} k_2 )),\\
k_4 &= f(t + \frac{12}{13} \Delta t, X + \Delta t (\frac{1932}{2197} k_1 - \frac{7200}{2197} k_2 + \frac{7296}{4104} k_3)),\\
k_5 &= f(t + \Delta t, X + \Delta t (\frac{439}{216} k_1 - 8k_2 + \frac{3680}{513} k_3 - \frac{845}{4104} k_4)),\\
k_6 &= f(t + \frac{1}{2} \Delta t, X + \Delta t (-\frac{8}{27} k_1 + 2k_2 - \frac{3544}{2565} k_3 + \frac{1859}{4104} k_4 - \frac{11}{40} k_5)).
\end{aligned}
$$

Dans la version minimale retenue ici, on utilise la combinaison de rang 4

$$
X_{n+1} = X_n + \Delta t (\frac{25}{216} k_1 + \frac{1408}{2565} k_3 + \frac{2197}{4104} k_4 - \frac15 k_5)
$$

garantissant une erreur locale de l’ordre de $\Delta t^5$.

````python
def rkf45_step(f, t, x, dt):
    """
    Effectue un pas d'intégration RKF-45 pour l'équation d'état X' = f(t, X).

    Paramètres
    ----------
    f : callable
        Fonction dérivant l'état X, signature f(t, X) -> dX/dt.
    t : float
        Temps courant (inutile ici car f ne l'utilise pas).
    x : ndarray
        Vecteur d'état au temps t.
    dt : float
        Pas de temps.

    Renvoie
    -------
    x_next : ndarray
        Vecteur d'état au temps t+dt estimé par le schéma RKF-45.
    """
    k1 = f(t, x)
    k2 = f(t + dt/4, x + dt/4 * k1)
    k3 = f(t + 3*dt/8, x + dt * (3/32*k1 + 9/32*k2))
    k4 = f(t + 12*dt/13, x + dt * (1932/2197*k1 - 7200/2197*k2 + 7296/4104*k3))
    k5 = f(t + dt, x + dt * (439/216*k1 - 8*k2 + 3680/513*k3 - 845/4104*k4))
    k6 = f(t + dt/2, x + dt * (-8/27*k1 + 2*k2 - 3544/2565*k3 + 1859/4104*k4 - 11/40*k5))
    return x + dt*(25/216*k1 + 1408/2565*k3 + 2197/4104*k4 - 1/5*k5)
````

### 4) Régulateur vectoriel PID 

Le booster devant être attrapé à vitesse quasi-nulle sur une position spécifique, il est impératif d'intégrer au code une boucle d'asservissement permettant de ramener simultanément, et à chaque instant, les erreurs vectorielles de position, de vitesse, d'accélération et d'altitude vers zéro. C'est la classe `PID`, pour **Proportionnel-Intégral-Dérivée**, qui instancie cette mécanique de correction de la trajectoire pour les phases de suicide et landing burn.

Ainsi, lorsque la fonction `__call__` reçoit une erreur $\boldsymbol e$ et un intervalle de temps $\Delta t$, elle met d’abord à jour la partie intégrale :  

$$\boldsymbol I \longleftarrow \boldsymbol I + \boldsymbol e \Delta t$$

Afin d'éviter l'overshooting, c'est-à-dire que le booster ait accumulé trop de vitesse lorsque la cible est atteinte (la dépassant donc) de telle sorte qu'il oscille indéfiniment autour, un seuil `windup_limit` est renseigné, saturant aussitôt l'intégale sur $\pm$ `windup_limit`. La composante dérivée est évaluée par simple différenciation finie :

$$\dot{\boldsymbol e} = \frac{\boldsymbol e - \boldsymbol e_{prec}}{\Delta t} \qquad \boldsymbol e_{prec} \leftarrow \boldsymbol e$$

Enfin, la sortie du correcteur s’exprime sous la forme 

$$\boldsymbol u = K_P \boldsymbol e + K_I \boldsymbol I + K_D \dot{\boldsymbol e}$$

où les gains $(K_P,K_I,K_D)$, que nous pouvons interpréter comme une récompense accordée au système pour avoir diminué ses erreurs, doivent être réglés durant la phase de test. Le **vecteur commande** $\boldsymbol u$ sert directement à moduler un accélérateur ou à alimenter un algorithme de poussée/gimbal.  

A chaque fois que le booster change de phase, les intégrales et historiques d'erreur sont effacés par le biais de la méthode `reset`, afin d'éviter de propager les biais d'une boucle à une autre. Finalement, grâce à la vectorisation via NumPy, les trois axes sont traités en parallèle, ce qui maintient un temps de calcul compatible avec la cadence de simulation.

<center><img src="PID.JPG" alt="Graphiques PUD" style="width:500px; height:600px;"></center>
<center>Figure 8 : Exemple de l'utilisation de méthodes PID dans le cadre d'un controleur de température.</center>

````python
class PID:
    """
    Contrôleur PID vectoriel à 3 composantes.

    Attributes
    ----------
    kp, ki, kd : float
        Gains proportionnel, intégral et dérivé.
    windup_limit : float or None
        Saturation de l'intégrateur pour éviter le windup.
    integral : ndarray, shape (3,)
        Terme intégral accumulé.
    prev_error : ndarray, shape (3,)
        Dernière erreur utilisée pour le terme dérivé.
    """
    def __init__(self, kp, ki, kd, windup_limit=None):
        self.kp, self.ki, self.kd = kp, ki, kd
        self.windup_limit = windup_limit
        self.integral = np.zeros(3)
        self.prev_error = np.zeros(3)

    def reset(self):
        """Réinitialise l'intégrateur et l'erreur précédente."""
        self.integral[:] = 0
        self.prev_error[:] = 0

    def __call__(self, error, dt):
        """
        Calcule la sortie PID pour une erreur vectorielle donnée.

        Parameters
        ----------
        error : ndarray, shape (3,)
            Erreur courante entre consigne et mesure.
        dt : float
            Pas de temps écoulé depuis l'appel précédent.

        Returns
        -------
        control : ndarray, shape (3,)
            Commande PID résultante.
        """
        self.integral += error * dt
        if self.windup_limit is not None:
            self.integral = np.clip(self.integral, -self.windup_limit, self.windup_limit)
        derivative = ((error - self.prev_error)/dt if dt>0 else np.zeros_like(error))
        self.prev_error = error.copy()
        return (self.kp*error + self.ki*self.integral + self.kd*derivative)
````

### 5) Modélisation des moteurs Raptor

La propulsion de SuperHeavy repose sur un ensemble de 33 réacteurs de nouvelle génération appelés Raptor. Afin de mener à bien notre modélisation, nous devons implémenter via la classe `RaptorGroup` les trois actions basiques que sont leur allumage, leur modulation ainsi que leur orientation : de cette manière, nous serons en mesure d'avoir accès à la poussée tridimensionnelle découlant d'une consommation de [propergols](https://fr.wikipedia.org/wiki/Propergol) donnée. Dans notre contexte, un maximum de 13 moteurs est utilisé, fidèlement à la configuration connue du véhicule : nous avons imposé pour plus de simplicité une poussée ainsi qu'un gimbaling, c'est-à-dire l'orientation des réacteurs en coordonnées sphériques ($\theta$, $\phi$), communs.

La poussée de chaque moteur, comprise entre 0 et la valeur maximale $T_{max}=2.26\times10^{6}$ N, est déterminée par la fonction `set_all_throttle` qui module la poussée maximale par un facteur compris entre 0 et 1. De son côté, la fonction `gimbal` stocke les deux angles sphériques ($\theta$, $\phi$), la longueur $r$ du Raptor étant constante, qui décrivent la déviation du jet de gaz par rapport à l'axe du booster?

Ces deux fonctions trouvent leur utilité physique dans `get_total_thrust`, qui permet de modéliser la poussée renvoyée sur chaque axe en coordonnées cartésiennes. Pour un réacteur braqué d'angles $\theta$, autour d'un axe horizontal, et $\phi$, atour de l'axe vertical, la projection en $(x, y, z)$ de la poussée s'écrit, en fonction de la poussée absolue $T_i = \text{throttle}_i \,T_\text{max}$ :

$$
\begin{aligned}
F_{x,i} &= T_i \sin\theta_i \cos\phi_i,\\
F_{y,i} &= T_i \sin\theta_i \sin\phi_i,\\
F_{z,i} &= T_i \cos\theta_i,
\end{aligned}
$$

Ces résultants de poussée $(F_x,F_y,F_z)$ s'ajouteront au bilan des forces aérodynamiques et gravitationnelles dans `Booster.f()`. Comme nous utilisons la vectorisation pour calculer et sommer ces composantes, nous pouvons nous permettre d'actualiser la poussée à chaque pas de 0.02s.
Finalement, afin d'implémenter la variation de masse totale du booster liée à la consommation du mélange LCH4-LOx (méthane liquide - oxygène liquide), nous avons ajouté la fonction `prop_flow`. Chaque Raptor fonctionnant à puissance maximale draîne $\dot{m_{full}} = 640$ kg/s du mélange LCH4-LOx, si bien que le débit global s'écrit, pour un `throttle` entre 0 et 1 donné :

$$\dot m_{\text{tot}} = (\sum_{i=1}^{n}\text{throttle}_i) \dot m_{\text{full}}$$

Cette quantité est immédiatement soustraite à la masse de propergol dans la classe `Booster` : ainsi, un braquage de gimbal qui compense un moment excentré est reflété instantanément dans les dynamiques linéaire massique.

<center><img src="raptor.jpg" alt="Réacteur Raptor" style="width:600px; height:350px;"></center>
<center>Figure 9 : Test d'un réacteur Raptor au centre de McGregor.</center>

````python
class RaptorGroup:
    """
    Groupe de Raptors vectorisé pour calculer pousée et consommation.
    Attributes
    ----------
    n : int
        Nombre de moteurs.
    max_thrust : float
        Poussée max par moteur (N).
    throttle : ndarray[n]
        Cadence de pousse [0..1].
    gimbal : ndarray[n,2]
        Angles de gimbal (theta, phi).
    flow_rate : float
        kg/s consommés par moteur à pleine puissance.
    """
    def __init__(self, n = 13, max_thrust = 2.26e6):
        self.n = n
        self.max_thrust = max_thrust
        self.throttle = np.zeros(n)
        self.gimbal = np.zeros((n,2))
        self.flow_rate = 640.0
        
    def set_all_throttle(self,val):
        """Fixe uniformément la throttle, en clipant [0,1]."""
        self.throttle[:] = np.clip(val,0,1)
        
    def set_all_gimbal(self,th,ph):
        """Fixe uniformément les angles de gimbal theta et phi."""
        self.gimbal[:] = th,ph
        
    def get_total_thrust(self):
        """
        Calcule la poussée totale (Fx, Fy, Fz) en sommant chaque moteur,
        compte tenu de la direction gimbal.
        """
        T = self.throttle*self.max_thrust
        th,ph = self.gimbal[:,0],self.gimbal[:,1]
        Fx = T*np.sin(th)*np.cos(ph)
        Fy = T*np.sin(th)*np.sin(ph)
        Fz = T*np.cos(th)
        return np.array([Fx.sum(),Fy.sum(),Fz.sum()])
    
    def prop_flow(self):
        """Flux massique total consommé (kg/s)."""
        return self.throttle.sum()*self.flow_rate
````

### 6) Modélisation aérodynamique

La descente d’un corps cylindrique, depuis 65 km jusqu’au sol, nous impose de traverser de façon continue les régimes subsonique, transsonique et supersonique : pour que notre intégrateur RKF-45 puisse suivre la dynamique sans discontinuités, il est nécessaire d'utiliser un modèle analytique à la fois léger en temps de calcul et parfaitement régulier sur tout le domaine de Mach et d’incidence concerné. La classe `Aerodynamics` encapsule cette logique en décrivant la traînée associée à la structure du booster, en ajustant automatiquement la surface "vue" par le flux d'air au fur et à mesure que l'orientation du booster varie, et en calculant les forces et couples produits par les quatre grid fins installées au sommet du corps principal. Tout d'abord, nous posons les bases suivantes :

- Le booster est assimilé à un cylindre. Cette géométrie maintient les ordres de grandeur de masse et d’inertie tout en rendant analytiques la surface frontale $A_{disk} = \pi r^{2}$ et le flanc latéral (cylindre) $A_{lat}=\pi d h$ ;

- Le vecteur `up`, que nous nommerons ici $\vec{u}$ et qui fixe l’orientation du booster, est obtenu en appliquant le quaternion d’attitude $q$ à la base canonique. De cette façon, la cinématique externe gérée par la classe `Booster` et le calcul aérodynamique restent parfaitement cohérents ;

- Dans la mesure où un corps cylindrique suivant une trajectoire majoritairement quasi-verticale génère peu de portance propre, la contribution $C_L^{structure}$ est posée nulle ;

- Les 4 grid fins, dont les aires sont égales à $5 \times 2.5 = 12.5$ m², sont supposées disposées tous les 90° autour du booster. Elles concentrent la portance utile et fournissent le moment de contrôle indispensable à la manoeuvre de descente.

````python
class GridFin:
    """
    Fixation de la position et de l'angle initial pour un grid-fin :
    - mount : position de montage
    - angle : angle initial
    """
    def __init__(self, mount, angle=0.0):
        self.mount=mount
        self.angle=angle
````

#### Force de traînée du booster

Dans notre contexte, la force de traînée liée à la structure du booster s’écrit  

$$\vec{F_D} = -\frac{1}{2} \rho(h) V^2 C_D(M) A_{eff}(\vec{v}, \vec{u}) \vec{v}$$  

où :

- $\rho(h)$ provient du modèle ISA ;

- $\vec{v}=\frac{\vec{v}}{V}$ est la direction de la vitesse du booster ;

- $V=\lVert\vec{v}\rVert$ sa norme ;

- $A_{eff}$ est la surface totale du booster que rencontre le flux d'air. Elle doit s’ajuster continûment tout au long de sa descente selon 

$$A_{eff} = A_{disk} \mu + A_{lat} \sqrt{1-\mu^{2}} \qquad \mu = \lvert \vec{v} \cdot \vec{u} \rvert$$ 

où $\vec{u}$ représente l'axe du booster, de sorte que lorsque $\mu\approx0$, la surface latérale domine, tandis que pour $\mu\approx1$, seule la section frontale subsiste. La racine carrée assure une transition lisse, sans changement brutal de pente ;

- $C_D(M)$ correspond au coefficient de traînée, calculé à partir du nombre de Mach. L'expérience montre que ce dernier présente, dans le cas d'un cylindre, un palier subsonique quasi constant, un pic transsonique centré sur $M=1$, une chute douce lorsque l’onde de choc s’est stabilisée, puis un plateau lorsque le régime supersonique s’installe. Pour capturer ce comportement sans créer la moindre cassure, nous adoptons le profil continu suivant :  

$$
C_D(M)=
\begin{cases}
C_{D0}, & M<0.8,\\[4pt]
C_{D0} + \dfrac{M-0.8}{0.4}\bigl(1-C_{D0}\bigr), & 0.8 \le M < 1.2,\\[8pt]
1 - 0.2\,\dfrac{M-1.2}{1.8}, & 1.2 \le M < 3,\\[8pt]
0.8, & M \ge 3,
\end{cases}
$$

avec $C_{D0}=0.3$. La pente est donc continue : aucune dérivée ne diverge, ce qui garantit la bonne tenue des PID qui s’appuient précisément sur ces dérivées.

````python
class Aerodynamics:
    """
    Modèle simplifié d'aérodynamique pour la fusée :
    - drag : force de traînée
    - lift_moment : portance et moment générés par les grid-fins
    """
    def __init__(self, atm, cd0 = 0.3, d = 9., h = 69., lf = 5., wf = 2.5):
        self.atm,self.cd0 = atm,cd0
        self.disk = np.pi*(d/2)**2
        self.cyl = np.pi*d*h
        self.Afin = lf*wf
        r,z = d/2,34.5
        ang = np.deg2rad([0,90,180,270])
        self.fin_pos = np.vstack([r*np.cos(ang), r*np.sin(ang), np.full_like(ang,z)]).T
        
    def cd(self, M):
        """
        Variation du coefficient de traînée en fonction du nombre de Mach.
        Linéaire de 0.3 à 1 entre M=0.8 et M=1.2, etc.
        """
        if M<.8:
            return self.cd0
        if M<1.2:
            return self.cd0+(M-.8)/.4*(1-self.cd0)
        if M<3.:
            return 1-(M-1.2)/1.8*0.2
        return 0.8
    
    def A_eff(self, v, up):
        """
        Surface efficace selon l'angle de vol.
        Combine surface disque et latérale.
        """
        V = np.linalg.norm(v)
        if V<1e-6:
            return self.disk
        mu = abs(np.dot(up,v/V))
        return self.disk*mu + self.cyl*np.sqrt(max(0.0,1-mu**2))
    
    def drag(self, v, alt, up):
        """
        Force de traînée aérodynamique
        """
        V = np.linalg.norm(v)
        if V<1e-3:
            return np.zeros(3)
        rho = self.atm.density(alt)
        D = 0.5*rho*V*V*self.cd(V/self.atm.sound_speed(alt))*self.A_eff(v,up)
        return -D*v/V
````

#### Rôle des grid fins

De son côté, la stabilité du booster est assurée par les grid fins. Pour une lame mince soumise à un angle $\alpha$ par rapport à son angle neutre, la théorie aérodynamique des profils minces, dans le cas linéaire, prédit un coefficient de portance $C_L = 2\pi \alpha (1-\tfrac{\alpha^{2}}{6})$, tandis que le coefficient de traînée est donné par $C_D(\alpha) = 0.01 + 0.10 \alpha^{2}$. Les forces exercées sur une grid fin portée à l’incidence $\alpha$ s’écrivent alors  

$$
\vec F_i = L_i\vec{l} - D_i \vec{v} \qquad
\begin{aligned}
L_i &= \tfrac12 \rho V^{2} A_{fin} C_L,\\
D_i &= \tfrac12 \rho V^{2} A_{fin} C_D
\end{aligned}
$$  

où le vecteur $\vec{l} = \frac{\vec{v} \times (\vec{v} \times \vec{u})}{\lVert \vec{v} \times (\vec{v} \times \vec{u})\rVert}$ est contenu dans le plan $(\vec{v}, \vec{u})$ et orthogonal à $\vec{v}$ : physiquement, la portance tombe donc naturellement à zéro si le flux d'air est orthogonal à la surface de la grid fin. Comme toutes les grid fins sont de positions fixées $\vec{r_i}$, alors elles livrent ensemble le couple total  

$$\vec{M_L} = \sum_{i=1}^{4} \vec{r_i} \times \vec{F_i}$$  

La fonction `lift_moment` retourne enfin le couple et la somme des forces :

$$(\vec{F_L}, \vec{M_L}) = (\sum_i \vec{F_i}, \vec{M_L}).$$  

Cependant, comme nous l'avons spécifié en début de paragraphe, nous utiliserons les grid fins à des fins de stabilité, et non d'orientation : en effet, une modélisation bien plus complexe des grid fins aurait été nécessaire pour simuler l'orientation fine nécessaire à la complétion de l'objectif de vol. Comme nous ne disposons ni des outils, ni de la puissance de calcul nécessaire, nous avons choisi de fixer les angles $\alpha$ des grid fins égaux à 0 : de cette sorte, nous ne bénéficions certes plus de leur portance, supposée induire des changements d'orientation, mais toujours de leur traînée, permettant au booster de ne pas basculer à l'envers. Toutefois, afin de conserver la dynamique d'orientation, nous avons décidé d'injecter manuellement les moments, supposément générés par la rotation des grid fins. De ce fait, nous évitons que le booster tourne librement sur lui même, tel une toupie.

````python
def lift_moment(self, v, alt, fin_ang, q):
        """
        Calcule la portance Fl et le moment Ml générés par les 4 grid-fins.
        - fin_ang : angles d'incidence vecteur de taille 4
        - q : quaternion d'orientation
        """
        V = np.linalg.norm(v)
        
        if V<1e-3:
            return np.zeros(3),np.zeros(3)
        rho = self.atm.density(alt)
        ref = np.cross(v,[0,0,1])
        
        if np.linalg.norm(ref)<1e-6:
            ref=np.cross(v,[0,1,0])
        ldir = np.cross(ref,v)
        ldir /= np.linalg.norm(ldir)
        vhat = v/V
        CL = 2*np.pi*fin_ang*(1-fin_ang**2/6)
        CD = 0.01+0.1*fin_ang**2
        
        if V/self.atm.sound_speed(alt)>1:
            CL *= 0.9
            CD *= 1.2
        Fl = 0.5*rho*V*V*self.Afin*CL
        Fd = 0.5*rho*V*V*self.Afin*CD
        F = Fl[:,None]*ldir - Fd[:,None]*vhat
        r = np.vstack([Quat.rot(p,q) for p in self.fin_pos])
        M = np.cross(r,F)
        return F.sum(0), M.sum(0)

Aerodynamics.lift_moment = lift_moment
````

### 7) Modélisation du booster

La classe `Booster` concentre la dynamique à 6 degrés de liberté (translation et rotation) du lanceur, ainsi que la consommation de propergols et la logique de changement de phase de vol. À chaque pas d’intégration $\Delta t$, elle évalue le système d’équations différentielles non linéaires

$$
\dot{\mathbf{X}} = f(t, \mathbf{X})
\qquad\text{avec}\qquad
\mathbf{X} \;=\;
\bigl[\,
\underbrace{\mathbf{r}}_{\text{pos}},\,
\underbrace{\mathbf{v}}_{\text{vel}},\,
\underbrace{q}_{\text{quat}},\,
\underbrace{\boldsymbol{\omega}}_{\text{rot}},\,
\underbrace{m_{\text{prop}}}_{\text{propergols}}
\bigr]^{\!\top}\in\mathbb{R}^{14},
$$

puis le fait évoluer au moyen de l'intégrateur RKF-45, garantissant une erreur locale d’ordre $\mathcal{O}\!(\Delta t^5)$ tout en restant assez léger pour que la boucle soit reproductible en temps réel.

#### Hypothèses

Nous effectuons les hypothèses suivantes :

- Le corps du booster est rigide, aucune déformation ou faiblesse structurelle n'étant prise en compte ;
- Le tenseur d'inertie $\mathbf{I}$ est ramené à celui d'un cylindre plein d'axe $z$ ;
- La masse de propergols $m_{prop}$ est uniformément répartie dans les réservoirs, permettant la mise à jour du tenseur d'inertie par simple proportionnalité de masse :

$$
\mathbf{I} = \operatorname{diag}\Bigl(\frac{1}{2}mr^{2}, \; \frac{1}{2}mr^{2}, \; \frac{m}{12}(3r^{2}+h^{2}) \Bigr),
\qquad
m=m_{\text{vide}}+m_{\text{prop}} .
$$

#### Forces et moments

Par le PFD nous obtenons :

$$\dot{\mathbf{v}} = \frac{\vec{P}+\vec{F_D}+\vec{F_L}+\vec{T}}{m}$$

où $\vec{P} = m \vec{g}$ est le poids du booster, sous l'hypothèse d'une force de gravité uniforme, $\vec{F_D}$ sa force de traînée, $\vec{F_L}$ la portance exercée sur les grid fins, et $\vec{T}$ la poussée des Raptors allumés. De son côté, la dynamique angulaire est donnée par l’équation d’Euler du corps rigide :

$$\mathbf{I} \, \dot{\vec{\omega}} = \vec{M_{cmd}} + \vec{M_{L}} - \vec{\omega} \times (\mathbf{I} \, \vec{\omega})$$

où $\vec{M_L}$ est le moment naturellement généré par la force de portance exercée sur les grid fins, qui dépend de la vitesse, de l'altitude, et de l'orientation du booster, ainsi que des angles d'incidence des grid fins. De son côté, $\vec{M_{cmd}}$ est le moment imposé, grâce à l'algorithme de pilotage (PID et SMC, voir ci-dessous), par les réacteurs Raptor pendant les deux dernières phases de descente.

L’attitude est, nous le rappelons, représentée par un quaternion unitaire $q$ de dérivée

$$\dot{q} = \frac{1}{2} \, q \otimes \Omega$$

où $\Omega = \bigl[0,\vec{\omega}\bigr] = (0, \omega_x, \omega_y, \omega_z) \in \mathbb{H}$ est un quaternion pur encodant la vitesse angulaire instantanée du booster.

````python
class Booster:
    """
    Modélisation complète du booster :
    - simulate f(t,X) = dX/dt
    - update_state : intègre X avec RKF-45
    - switch de phase primaire → secondary → tertiary
    """
    def __init__(self, atm, aero, pos0, vel0, target):
        self.atm, self.aero = atm, aero
        self.empty = 275e3
        self.prop  = 200e3
        self.pos = pos0.astype(float)
        self.vel = vel0.astype(float)
        self.q   = np.array([1.,0.,0.,0.])
        self.w   = np.zeros(3)
        self.fins= [GridFin(0.) for _ in range(4)]
        self.rap13 = RaptorGroup(13)
        self.rap3  = RaptorGroup(3)
        self.phase = Phase.PRIMARY
        self.target= target

        # cascade PID pour la phase tertiary
        self.pid_pos = PID(kp=0.08, ki=0.01, kd=0.04, windup_limit=300.) #0.02, 0.005, 0.01, 100.0
        self.pid_vel = PID(kp=1.0 , ki=0.10, kd=0.4 , windup_limit=100.) #0.5, 0.05, 0.2, 50.0

        self.vz_ref    = -80.0
        self.current_dt= 0.1

    def inertia(self):
        """
        Calcule la matrice d'inertie approximée (cylindre + disque).
        """
        r,h = 4.5, 69.0
        m = self.empty + self.prop
        return np.diag([0.5*m*r*r,
                        0.5*m*r*r,
                        m*(3*r*r+h*h)/12])
````

#### Boucles d’asservissement et phases de vol

Le déclenchement des 3 phases de descente, énumérées dans la classe `Phase`, est initié dans la classe `Booster` selon les conditions suivantes :

- La phase de descente contrôlée (`PRIMARY`) est déclenchée dès le départ de la simulation. A cette étape, les moteur sont coupés, et seules les actions des grid fins permettent d'orienter correctement le booster : afin de coordonner les rotations des grid fins de manière cohérente, nous utilisons une loi de pilotage non linéaire appelée Sliding Mode Control (ou SMC), guidant le booster vers sa zone d'atterrissage ; 


- La transition entre les première et deuxième (suicide burn, `SECONDARY`) phases s'établit à une altitude de 1500m, conformément à la réalité ;


- le déclenchement de la troisième et dernière phase (landing burn, `TERTIARY`) s'effectue à une vitesse maximale de 80m/s. Ici, une double boucle PID est utilisé afin de neutraliser les erreurs de position et de vitesse.

````python
class Phase(Enum):
    PRIMARY=1; SECONDARY=2; TERTIARY=3
````

##### a) Descente contrôlée et Sliding Mode Control

Le Sliding Mode Control repose sur le suivi d'une surface de glissement, dont le vecteur associé $\vec{s}$ encode l'écart à la trajectoire cible. Une fois cette surface définie, des commandes sont envoyées aux différentes composantes du booster, et ce afin d'imposer une dynamique de "glissement" sur cette surface. Dans notre cas, le SMC est utilisé dans deux boucles distinctes de la classe `Booster`, exclusivement en phase `PRIMARY` : nous avons choisi cette méthode en raison de sa réactivité, car nécessitant un faible temps de calcul, ainsi que de sa simplicité d'implémentation.

Tout d'abord, le contrôle de translation `smc_transl()` a pour objectif la convergence finale du centre de masse du booster vers des positions et vitesses de référence : 

$$\vec{{r}_{ref}} = (x_{cible}, y_{cible}, 1500\;\text{m}),
\qquad
\vec{{v}_{ref}} = (0, 0, -300)\;\text{m/s}$$

La vitesse de référence a été choisie par rapport aux données réelles de télémétrie, tandis que l'altitude cible a été fixée à 1500m, seuil de transition avec la phase suivante.
Le vecteur de glissement, en tant qu'évaluateur de l'écart entre trajectoires cible et réelle, est défini par :

$$ \vec{s} = (\vec{v} - \vec{{v}_{ref}}) + \lambda(\vec{r} - \vec{{r}_{ref}})$$

où $\vec{r}$ est la position actuelle du booster, $\vec{v}$ sa vitesse et $\lambda > 0$ un gain d’amortissement, déterminant la "brutalité de la convergence. De cette expression, nous déduisons qu'un suivi idéal de la surface de glissement implique une nullité constante du vecteur $\vec{s}$, et ce à toute étape de la descente contrôlée. La loi de commande d’accélération associée à $\vec{s}$ est alors :

$$\boxed{\vec{{a}_{cmd}} = -K \cdot \tanh(\frac{\vec{s}}{\phi})}$$

où le gain $K$ règle la vitesse de convergence et le tableau $\vec{\phi}$ permet de normaliser les composantes de $\vec{s}$ avant l'application de la tangente hyperbolique, ce qui adoucit ou durcit la réponse du contrôleur selon l'axe. Comme les réacteurs Raptor sont éteints, cette accélération n'est pas directement convertie en poussée, mais possède plutôt un rôle d'objectif directionnel, permettant de déterminer par la suite l'orientation idéale du booster.

C'est sur cette dimension qu'agit la fonction `smc_att()`, en implémentant un contrôleur SMC spécialement conçu pour l'attitude du booster. Cette dernière, représentée par les angles d'Euler $(\theta, \psi, \phi) =$ (tangage, lacet, roulis), doit idéalement converger vers une consigne dynamique calculée par la fonction `att_primary()`. Dans ce cadre, le vecteur de glissement devient :

$$\vec{s} = (\vec{\theta} - \vec{{\theta}_{ref}} \; , \; \vec{\psi} - \vec{{\psi}_{ref}} \; , \; \vec{\phi} - \vec{{\phi}_{ref}}) + \lambda(-\vec{\omega})$$

où $\vec{\omega}$ est la vitesse angulaire du booster dans son propre repère. Pour des raisons de simplification notables, nous avons préféré laisser libre l'angle $\phi$, c'est-à-dire de roulis : en effet, la structure du booster étant symétrique selon l'axe de roulis, une rotation autour de ce dernier n'a aucune incidence sur la trajectoire.

Finalement, la commande agissant sur le moment est ici choisie discontinue, de sorte à induire une commande rapide :

$$\boxed{\vec{M}_{\text{cmd}} = -k \cdot \operatorname{sign}(\vec{s})}$$

````python
def att_primary(self):
    """
    Loisir d'attitude en phase primaire selon altitude (pitch)
    + orientation vers la target (yaw).
    """
    z = self.pos[2]
    if   z>65e3: pitch=0
    elif z>40e3: pitch=.785*(1-(z-40e3)/25e3)
    elif z>3e3:  pitch=.785
    elif z>2e3:  pitch=.785*(z-2e3)/1e3
    else:        pitch=0
    yaw = np.arctan2(-self.pos[1], -self.pos[0])
    return pitch, yaw, 0

def smc_transl(self):
    """
    Sliding-mode control pour la translation en phase primaire.
    Cible position 3000m au-dessus de la target, v_z=-300 m/s.
    """
    ref_p = np.array([*self.target[:2], 1500.0])
    ref_v = np.array([0.,    0., -300.0])
    s     = (self.vel - ref_v) + 0.5*(self.pos - ref_p)
    K     = np.array([.5,.5,.3])
    phi   = np.array([20,20,10])
    u     = -K*np.tanh(s/phi)
    u[:2] = np.clip(u[:2], -5, 5)
    u[2]  = np.clip(u[2], -3, 3)
    return u

def smc_att(self, dp, dy, dr):
    """
    Sliding-mode control pour l'attitude (pitch, yaw, roll).
    dp, dy, dr : angles désirés.
    """
    p, y, r = self.euler()
    # L'angle de roulis est laissé libre
    s = np.array([dp-p, dy-y, 0]) + 0.5*(-self.w)
    return -1e4 * np.sign(s)

def euler(self):
        """
        Convertit quaternion en angles Euler (pitch, yaw, roll).
        """
        w, x, y, z = self.q
        yaw  = np.arctan2(2*(w*z + x*y), 1 - 2*(y*y+z*z))
        sinp = 2*(w*y - z*x)
        pitch= np.pi/2 if abs(sinp)>=1 else np.arcsin(sinp)
        roll = np.arctan2(2*(w*x + y*z), 1 - 2*(x*x+y*y))
        return pitch, yaw, roll

Booster.att_primary = att_primary
Booster.smc_transl = smc_transl
Booster.smc_att = smc_att
Booster.euler = euler
````

##### b) Suicide burn

Une fois l'altitude de transition atteinte, à savoir $z - z_{cible} \leq 1500$m, le booster entre en phase `SECONDARY`, associée à un freinage brutal par le biais d'un rallumage rapide de 13 de ses réacteurs Raptor. L'objectif étant de compenser au plus vite l'énergie cinétique verticale du booster, les réacteurs sont allumés à **pleine puissance** selon la commande 

```python
self.rap13.set_all_throttle(1.0)
self.rap13.set_all_gimbal(0.0, 0.0)
```

Cela signifie que le throttle, c'est-à-dire la modulation de la puissance maximale par un facteur compris entre 0 et 1, est fixé à 1, tandis que le gimbal, correspondant à l'orientation en coordonnées sphériques des réacteurs, est neutralisé afin de fournir un poussée parallèle à l'axe du booster. Comme il s'agit d'une phase courte et violente, nous avons préféré n'imposer aucune loi d'asservissement translationnel ou rotationnel : en effet, le sens de la poussée suffit à lui seul à neutraliser une grande partie de la vitesse verticale.

##### c) Landing burn et PID

Lorsque la vitesse verticale devient inférieure à 80m/s, la troisième et dernière phase se déclenche : celle-ci ayant pour unique objectif d'optimiser précisément la position et la vitesse du booster avant son atterrissage lors de l'atterrissage, il semble nécessaire de recourir à une double boucle d'asservissement de type PID. Nous en avons ici défini deux types :

- le PID lié à la correction de la position du booster (`pid_pos`), générant une vitesse de signe opposé à l'erreur sur la position, définie comme :

$$\vec{e}_r = \vec{r}-\vec{r}_{\text{cible}}$$

- le PID lié à la correction de la vitesse du booster (`pid_vel`), calculant l'accélération requise pour contrer l'erreur sur la vitesse, définie comme :

$$\vec{e}_v = \vec{v}-\vec{v}_{\text{ref}}$$

De ces deux équations, nous pouvons obtenir les vitesses de référence ainsi que les commandes d'accélération en utilisant :

$$\boxed{\vec{v^{\star}} = -\text{PID}_{\text{pos}}\!(\vec{r} - \vec{r_{cible}})},$$
$$\boxed{\vec{a}_{\text{cmd}} = -\text{PID}_{\text{vel}}(\vec{v} - \bigl[-\text{PID}_{\text{pos}}\!(\vec{r} - \vec{r_{cible}}) \bigr]) = -\text{PID}_{\text{vel}}(\vec{v} - \vec{v^{\star}})}$$

avec $\text{PID}(\vec{e}) = K_P\vec{e}+K_I\!\!\int\!\vec{e}\,dt+K_D\frac{d\vec{e}}{dt}$.

Finalement, pour maintenir un vol stationnaire au voisinage du sol (nécessaire pour se faire attraper par les chopsticks), une correction gravitationnelle $(0,0,+g)$ est ajoutée à la composante verticale de l’accélération $a_{\text{cmd},z}$, et ce dès que l’altitude résiduelle $|z-z_{\text{cible}}|$ passe sous la barre des 50 m. Nous avons ainsi :

$$
\vec{a}_{\text{cmd}} =
\begin{cases}
-\text{PID}_{\text{vel}}(\vec{v} - \vec{v}^{\star})
& \text{si } z > 50\,\text{m} \\
-\vec{g} -\text{PID}_{\text{vel}}(\vec{v} - \vec{v}^{\star}) & \text{sinon}
\end{cases}
$$

Une fois l'accélération cible obtenue, elle est convertie en force requise

$$\vec{F}_{\text{req}} = m \cdot \vec{a}_{\text{cmd}}$$

dont la norme détermine le throttle total pour les 3 réacteurs Raptor $\tau = \frac{|\vec{F}_{\text{req}}|}{3T_{\max}}\;\bigl[0,1\bigr]$, et la direction fixe leurs angles de gimbal $(\theta,\phi)$ via une conversion sphérique :

$$\theta = \arccos\left(\frac{F_z}{\|\vec{F}_{\text{req}}\|}\right),\qquad\phi = \arctan2(F_y,F_x)$$

Ces deux paramètres sont transmis aux moteurs par :

```python
self.rap3.set_all_throttle(tau)
self.rap3.set_all_gimbal(theta, phi)
```

##### d) Gestion de la masse propulsive

Pour chaque groupe de moteurs impliqué lors des tirs de suicide et landing burn, nous devons mettre à jour à chaque instant la quantité de propergols consommés. Ce processus est impératif, car la masse du booster a une incidence non négligeable sur le throttle : en effet, une plus faible masse nécessite moins de poussée pour ralentir. La masse de propergols suit donc la loi suivante :

$$\dot m_{\text{prop}} = -\sum_{i=1}^{n}\tau_i\,\dot m_{\max}
\quad\Longrightarrow\quad
m_{\text{prop}}(t+\Delta t)=\max\bigl(0,\;m_{\text{prop}}+\dot m_{\text{prop}}\Delta t\bigr),
$$

avec $\dot m_{\max}=640$ kg s$^{-1}$ par Raptor. Lorsque le carburant est épuisé, la poussée est donc automatiquement forcée à zéro, assurant la cohérence du modèle.

### 8) Ordinateur de bord

À chaque itération, l’ordinateur de bord exécute un cycle discret de durée $\Delta t$ comprise entre 0.02s (lorsque l'altitude tombe sous 200m) et 0.1s (par défaut).
Pour commencer, le vecteur d'état 

$$\mathbf X = \bigl[\;
\underbrace{\vec r}_{\text{pos}},
\underbrace{\vec v}_{\text{vel}},
\underbrace{q}_{\text{quat}},
\underbrace{\boldsymbol\omega}_{\text{rot}},
\underbrace{m_{\text{prop}}}_{\text{ergols}}
\bigr]^{\!\top}\!
\in\mathbb R^{14}
$$

est construit à partir des variables internes du booster, à savoir sa position, sa vitesse, son orientation, sa vitesse angulaire, et la masse de propergols qu'il contient. De son côté, puisque la masse est vouée à évoluer, le tenseur d'inertie $\mathbf I(m)$ est lui aussi calculé.
Ensuite, la fonction `Booster.f`, dont l'objectif est de représenter le système dynamique complet du booster, est appelée : elle utilise ces données afin de calculer, en s'appuyant sur les classes associées, les forces aérodynamiques, le poids et la poussée des Raptor, puis sonde le booster afin de déterminer la phase qu'il est en train de traverser. Selon le résultat (`PRIMARY`, `SECONDARY` ou `TERTIARY`), différentes commandes d'accélération et de modification d'orientation sont injectées dans la simulation. Par exemple, si le booster est en phase `PRIMARY`, l'ordinateur de bord lance la loi de pilotage SMC, contrôlant l'orientation du booster lorsque sa poussée est nulle ; tandis que, dans le cas où le booster est en phase `TERTIARY`, c'est la double boucle PID qui est appelée, afin de moduler l'orientation (donc la position et la vitesse) du booster par le biais des réacteurs Raptor.

De là, la méthode `Booster.f` renvoie :

- $\dot{\vec r}=\vec v$,
- $\dot{\vec v}= \vec F_{tot}/m$,
- $\dot q = \tfrac12\,q\otimes[0,\boldsymbol\omega]$,
- $\mathbf I\dot{\boldsymbol\omega} = \vec M_{tot} - \boldsymbol\omega\times(\mathbf I\boldsymbol\omega)$,
- $\dot m_{\text{prop}} = -\dot m(\tau)$.

````python
def f(self, t, X):
    """
    Terme de droite de l'EDO d'état X = [pos, vel, q, w, prop_mass].
    Retourne dX/dt = [vel, acc, dq, wdot, prop_dot].
    """
    pos, vel, q, w, prop = X[:3], X[3:6], X[6:10], X[10:13], X[13]
    m = self.empty + prop
    up= Quat.rot([0,0,1], q)

    Fd = self.aero.drag(vel, pos[2], up)
    Fl, Ml = self.aero.lift_moment(
        vel, pos[2],
        np.fromiter((f.angle for f in self.fins), float),q)

    cmd_acc = np.zeros(3)
    thrust  = np.zeros(3)
    M_cmd   = np.zeros(3)
    prop_dot= 0.0

    if self.phase is Phase.PRIMARY:
        a_cmd = self.smc_transl(); a_cmd[2]=0
        thrust   = np.zeros(3)
        M_cmd    = self.smc_att(*self.att_primary())
        prop_dot = 0.0

    elif self.phase is Phase.SECONDARY:
        # suicide burn full throttle
        self.rap13.set_all_throttle(1.0)
        self.rap13.set_all_gimbal(0.0, 0.0)
        thrust   = self.rap13.get_total_thrust()
        prop_dot = -self.rap13.prop_flow()

    else:  # TERTIARY
        pos_err = pos - self.target
        vel_cmd_cascade = -self.pid_pos(pos_err, self.current_dt)
        v_cmd = vel_cmd_cascade  # hover
        vel_err = vel - v_cmd
        cmd_acc = -self.pid_vel(vel_err, self.current_dt)
        if abs(pos_err[2]) < 50.0:
            cmd_acc[2] += 9.81

        F_req    = m * cmd_acc
        F_req_mag= np.linalg.norm(F_req)
        RG, nR   = self.rap3, 3
        Tmax_tot = nR*RG.max_thrust
        thr      = np.clip(F_req_mag/Tmax_tot,0,1) if F_req_mag>0 else 0.0
        RG.set_all_throttle(thr)
        if F_req_mag>1e-6:
            tv=np.clip(F_req/F_req_mag,-1,1)
            th=np.arccos(tv[2]); ph=np.arctan2(tv[1],tv[0])
            RG.set_all_gimbal(th,ph)
        thrust   = RG.get_total_thrust()
        prop_dot = -RG.prop_flow()

    Fg  = np.array([0,0,-9.81*m])
    acc = (Fd + Fl + thrust + Fg)/m

    dq  = 0.5 * Quat.mul(q, np.hstack(([0.], w)))
    wdot= np.linalg.inv(self.inertia()) @ (Quat.rot(M_cmd+Ml, q) - np.cross(w, self.inertia()@w))

    return np.hstack((vel, acc, dq, wdot, [prop_dot]))

Booster.f = f
````

Finalement, la fonction `update_state`, dont la mission est de faire avancer l'évolution dynamique du booster d'un pas de temps $\Delta t$, entre en jeu. A partir du vecteur $X_n$, fourni en sortie de `Booster.f`, `update_state` appelle l'intégrateur RKF-45 à pas fixe via sa fonction `rkf45_step` : de cette façon, le prochain état $X_{n+1}$ est déterminé, le permettant d'être à son tour utilisé pour mettre à jour la dynamique du booster au fur et à mesure du temps.

<center><img src="booster_landing.jpeg" alt="Booster landing burn" style="width:1200px; height:500px;"></center>
<center>Figure 11 : Superposition d'images du landing burn. <br> La dynamique de correction basée sur l'orientation est clairement visible.</center>

````python
def update_state(self, dt):
    """
    Intègre l'état d'un pas dt avec RKF-45, met à jour pos, vel, q, w, prop.
    Gère aussi les transitions de phase.
    """
    if self.pos[2]<200: dt=0.02
    self.current_dt = dt
    X  = np.hstack((self.pos, self.vel, self.q, self.w, [self.prop]))
    Xn = rkf45_step(self.f, 0, X, dt)
    self.pos,self.vel = Xn[:3], Xn[3:6]
    self.q,self.w     = Quat.norm(Xn[6:10]), Xn[10:13]
    self.prop         = max(Xn[13],0.0)

    if self.phase is Phase.PRIMARY and self.pos[2]-self.target[2]<=1500:
        self.phase = Phase.SECONDARY
        self.pid_pos.reset(); self.pid_vel.reset()
    if self.phase is Phase.SECONDARY \
    and abs(self.vel[2])<abs(self.vz_ref):
        self.phase = Phase.TERTIARY
        self.pid_pos.reset(); self.pid_vel.reset()
        
Booster.update_state = update_state
````

## III. Lancement de la simulation et résultats

Les lignes de codes ci-dessous sont dédiées au lancement de la simulation et du stockage de la télémétrie sous la forme :
- d'un suivi en direct de la simulation, idéal pour une première analyse du déroulement de la modélisation ;
- d'un fichier de logs en format .CSV, interprétable par la librairie `pandas` ;
- de graphes, dont l'étude est utile pour une analyse visuelle est intégrale de chaque variable utilisée.

Nous initialisons la position du booster à une altitude de 65km, et la plaçons aux coordonnées $(1000.0, -100.0)$m sur le plan $(x, y)$ : sa vitesse de départ est fixée à $(v_x = -20.0, v_y = 1.0, v_z = -75.0)$, tandis que sa position d'atterissage correspond aux coordonnées $(0.0, 0.0, 90.0)$.
Plutôt que d'imposer comme condition de fin de simulation le rejoignement de la cible par le booster, n'impliquant pas nécessairement une vitesse quasi-nulle, nous en avons choisi deux :

- La norme du vecteur calculant l'erreur de position doit être inférieure ou égale à 1m, c'est-à-dire

$$\sqrt{(x - x_{cible})^2 + (y - y_{cible})^2 + (z - z_{cible})^2} \leq 1 \; \text{m} \; ;$$

- La norme du vecteur vitesse du booster doit être inférieure ou égale à 2 m/s, soit

$$\sqrt{(v_x^2 + v_y^2 + v_z^2)} \leq 2 \; \text{m/s}.$$

Nous avons choisi ces valeurs relativement aux dimensions du booster et des chopsticks, telles qu'une erreur de position totale de 1m ne porte pas préjudice au bon déroulement de l'atterrissage, et à la vitesse réelle d'atterrissage du booster, avoisinant les 10 km/h. En effet, dans la pratique, la position ainsi que la vitesse angulaire de fermeture des bras mécaniques sont calculés en fonction de la déviation latérale du booster : de cette manière, dans le cas où le booster ne venait qu'à rencontrer une difficulté de correction de son erreur latérale, il pourrait tout de même être récupéré. De son côté, la vitesse verticale "élevée" à l'atterrissage est amortie par des rails mobiles de part et d'autre des chopsticks, permettant au booster de ne pas utiliser le carburant qui aurait été nécessaire à annuler intégralement sa vitesse. 


Afin d'améliorer la lisibilité des données, la télémétrie en direct est affichée toutes les 5 secondes ; les graphes ainsi que le fichier de télémétrie peuvent, eux, être consultés dès la fin de la simulation, naturelle ou forcée.

````python
if __name__=="__main__":
    atm   = Atmosphere()
    aero  = Aerodynamics(atm)
    booster = Booster(
        atm,aero,
        pos0   = np.array([1000.0, -100.0, 65000.0]),
        vel0   = np.array([-20.0, 1.0, -875.0]),
        target = np.array([0.0, 0.0, 90.0])
    )

    log = {k:[] for k in ["t","alt","rho","x","y","vx","vy","vz",
                          "mass","pitch","yaw","roll","throttle",
                          "drag","lift"]}
    dt, t = 0.1, 0.0
    try:
        while np.sqrt((booster.pos[0] - booster.target[0])**2 + (booster.pos[1] - booster.target[1])**2 +
                      (booster.pos[2] - booster.target[2])**2) > 1 or np.sqrt(booster.vel[0]**2 + booster.vel[1]**2 + booster.vel[2]**2) > 2:
            
            up = Quat.rot([0,0,1], booster.q)
            Fd = aero.drag(booster.vel, booster.pos[2], up)
            fins_ang = np.fromiter((f.angle for f in booster.fins),float)
            Fl,_ = aero.lift_moment(booster.vel, booster.pos[2], fins_ang, booster.q)
            
            log["t"].append(t)
            log["alt"].append(booster.pos[2])
            log["rho"].append(atm.density(booster.pos[2]))
            log["x"].append(booster.pos[0]); log["y"].append(booster.pos[1])
            log["vx"].append(booster.vel[0]); log["vy"].append(booster.vel[1])
            log["vz"].append(booster.vel[2])
            log["mass"].append(booster.empty+booster.prop)
            
            p,y,r=booster.euler()
            log["pitch"].append(np.degrees(p))
            log["yaw"].append(np.degrees(y))
            log["roll"].append(np.degrees(r))
            
            thr = (1.0 if booster.phase is Phase.SECONDARY
                   else booster.rap3.throttle.mean()
                   if booster.phase is Phase.TERTIARY else 0.0)
            log["throttle"].append(thr)
            
            log["drag"].append(np.linalg.norm(Fd))
            log["lift"].append(np.linalg.norm(Fl))

            booster.update_state(dt)
            t+=dt
            if int(t*10)%50==0:
                print(f"t={t:5.1f}s  {booster.phase.name:<9}"
                      f"  x={booster.pos[0]:7.2f}"
                      f"  y={booster.pos[1]:7.2f}"
                      f"  z={booster.pos[2]:7.2f}"
                      f"  Vx={booster.vel[0]:7.2f}"
                      f"  Vy={booster.vel[1]:7.2f}"
                      f"  Vz={booster.vel[2]:7.2f}")
    except KeyboardInterrupt:
        print("Simulation interrompue.")

    pd.DataFrame(log).to_csv("telemetrie_booster.csv",index=False)
    print("CSV écrit.")
````

### 1) Position et vitesse

Les forces aérodynamiques ainsi que le pilotage SMC ont les impacts suivants sur la position et la vitesse du booster :
- En phase transsonique, c'est-à-dire autour de Mach 1, le coefficient de traînée $C_D$ rencontre un pic, d'où un "mur" de traînée à cette phase de vol. Toutefois, la valeur de $C_D$ reste moins influente que les valeurs de vitesse (la force de traînée est proportionnelle à la vitesse au carré) et de densité atmosphérique $\rho$ dans le ralentissement ;
- En phase `PRIMARY`, le booster s'oriente activement via le Sliding Mode Control (SMC) pour adopter l'attitude offrant une surface efficace $A_{eff}$ maximale (donc une traînée maximale, comme elle est proportionnelle à la surface). En fin de phase, la vitesse du booster est donc "faible", conformément à la télémétrie réelle. En pratique, l'énergie cinétique est convertie en chaleur (voir photo ci-dessous), ce que nous avons choisi de négliger ici car cela n'apportait rien au modèle.

Nous constatons qu'elle dure à peu près moitié moins de temps (environ 2s) que dans la réalité (4-5s) : cela est très cohérent, car la vitesse à l'allumage est deux fois plus faible que durant les vols officiels, en majeure partie à cause des nos hypothèses de simplification. Celle relative à l'absence de déformation de la structure du booster, en l'occurrence, permet à ce dernier de supporter des forces de traînée bien plus élevée dans notre simulation : en réalité, un angle d'approche trop agressif à basse altitude et haute vitesse engendrerait une déformation/casse du booster, ainsi qu'une fonte des réacteurs Raptor à cause de la transformation trop brutale de l'énergie cinétique en énergie thermique ;
    
- Durant le landing burn, le booster réduit progressivement sa vitesse $v_z$ jusqu'à la cible en $z$. Lorsqu'il l'atteint, il se met en mode "hover" : il tente de fournir la poussée nécessaire pour compenser la gravité, tout en se ramenant latéralement vers la cible. Une fois la cible atteinte dans l'espace, on constate un comportement parasite typique des boucles d'asservissement PID : le booster oscille autour de sa position d'équilibre, sans jamais vraiment s'équilibrer autour. Cette manoeuvre est donc d'efficacité très limitée : à cause des oscillations induites par le PID et une correction de la position pas assez brutale, sa durée est beaucoup trop longue (215s contre 20s en réalité). Pourtant, au bout de 25-30s de landing burn, nous constatons que le booster arrive à altitude cible avec une erreur latérale certes non néglieable, mais modérément éloignée des chopsticks : il s'agit d'une bonne performance, bien que les oscillations respectives autour de $x_{cible}$, $y_{cible}$ et $z_{cible}$ engendrent des erreurs de position et de vitesse ayant du mal à être minimisées conjointement (comme la condition d'arrêt de la simulation est un "or", il faut que les deux conditions soient remplies simultanément). Nous constatons toutefois un point positif notable : la dynamique d'évolution du carburant est fidèle à la réalité, avec une quantité suffisante de carburant pour toute la durée du vol ; en réalité, cependant, nous chercherions plutôt à réduire au maximum la durée du landing burn afin d'économiser le plus de carburant possible. Bien que le PID soit un modèle très simplifié, présentant des limites notables, nous n'aurions pas pu faire mieux : nous avons essayé d'utiliser le SMC et une autre méthode appelée Model Predictive Control (qui prédit l'évolution du système sur un horizon de temps donné, et optimise les commandes de sorte à se conforter ou non dans cette prédiction), mais toutes deux présentaient des oscillations plus amples. D'autres méthodes plus robustes existent, mais sont très coûteuses en temps de calcul : nous étions donc limités par la puissance de calcul et le langage de programmation, globalement inadaptés à des manoeuvres extrêmement précises comme le landing burn.

### 2) Orientation

Plus tôt, nous avons fait mention des angles d'Euler $(\theta, \psi, \phi)$, appelés tangage, lacet et roulis dans le jargon aérospatial. Puisque ces notions peuvent paraître abstraites au premier coup d'oeil, prenons le temps, dans l'optique finale d'évaluer la verticalité du booster à l'atterrissage, de comprendre la correspondance de ces rotations. Pour cela, nous allons utiliser le repère des coordonnées sphériques $(r, \theta, \phi)$, avec lequel l'analogie est quasiment directe. 

Considérons notre booster de longueur $\rho$, dont la base se situe en l'origine $O$ et le sommet en $P$, de coordonnée $\rho \vec{u_r}$, et tel qu'orienté d'angles $\theta$ par rapport à la verticale $z$ et $\phi$ par rapport à $x$.

<center><img src="spherique.webp" alt="Coordonnées sphériques" style="width:400px; height:400px;"></center>
<center>Figure 13 : Notre booster s'étend sur le segment [OP].</center>

Dans ce système de coordonnées, les trois rotations associées aux angles d'Euler sont les suivantes :

- Le roulis peut être associé à une rotation du booster sur lui-même, c'est-à-dire autour de l'axe $[OP]$. Puisqu'il n'est d'aucune utilité aérodynamique et qu'il ne nous renseigne en rien sur la verticalité du booster, nous avions choisi de le laisser libre ;
- La modification du tangage correspond à une augmentation ou diminution de l'angle $\theta$. C'est cet angle qui joue sur la verticalité à proprement parler du booster, en l'éloignant ou le rapprochant de l'axe $z$ ;
- Le lacet, lui, traduit une rotation autour de l'axe $z$, c'est-à-dire un changement d'angle $\phi$. Bien qu'il ne soit en rien lié à la verticalité du booster, il est utile en complément avec le tangage pour l'orienter vers sa cible.

````python
# Graphes d'orientation
fig3, ax3 = plt.subplots(2,1,figsize=(10,10))
ax3[0].plot(log["t"], log["pitch"]); ax3[0].set_title("Tangage en fonction du temps"); ax3[0].set_ylabel("Angle de tangage"); ax3[0].grid()
ax3[1].plot(log["t"], log["yaw"]); ax3[1].set_title("Lacet en fonction du temps"); ax3[1].set_ylabel("Angle de lacet"); ax3[1].grid()
for a in ax3:
    a.set_xlabel("Temps (s)")
plt.tight_layout()
plt.show()
````

Nous observons, à partir des données de télémétrie de la simulation, que le tangage en fin de vol du booster est de $\theta \approx 2$° : bien que cet angle soit non nul et ne permette pas de conclure sur une parfaite verticalité, il est impératif de se rapprocher des ordres de grandeur. En effet, considérons le triangle rectangle dont l'hypoténuse correspond à la longueur du booster, et les deux autres longueurs aux composantes verticale ($z$) et horizontale ($x$ ou $y$) du vecteur $\vec{OP}$ : la composante horizontale est alors de norme

$$\rho \sin(\theta) = 69 \times \sin(2°) = 2.4 \; \text{m}.$$

Aux vues de la hauteur ainsi que du diamètre du booster, nous pouvons évidemment négliger cette composante : il est par ailleurs courant, dans l'industrie aérospatiale, de considérer le tangage nul lorsqu'il est inférieur ou égal à 2°, tant cet angle est quasiment imperceptible devant les dimensions des véhicules.
L'objectif de verticalité à l'atterrissage est donc bel et bien rempli. Cependant, afin d'envisager l'utilisation des chopsticks, qui correspondent aux bras mécaniques ayant attrapé le booster en plein vol, il est nécessaire de considérer un degré de liberté supplémentaire : celui du roulis, permettant en l'occurrence d'orienter le booster de sorte à ce que ses supports d'atterrissage soient parfaitement alignés avec les chopsticks.
